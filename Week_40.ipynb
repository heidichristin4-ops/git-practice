```python
# Imports: 

import pandas as pd
import yaml

# Leser YAML-filen
with open("config.yml", "r") as file:
    yml_data = yaml.safe_load(file)

# Konverterer til Pandas DataFrame
df = pd.DataFrame([yml_data])

#print(df)

# Leser Xlsx-filen
df_sensors = pd.read_excel("sensors.xlsx")
#print(df_sensors)

# Leser CSV-filen
df_csv = pd.read_csv("calibrations.csv")
#print(df_csv)

# Matcher hver sensor plassering med kalibreringsdataene
merged_df = pd.merge(df_sensors, df_csv, on="sensor_id", how="left")
print(merged_df)

#filtrerer days_since_calibration > max_days_since_calibration  
filtered_df = merged_df[merged_df["days_since_calibration"] > yml_data["max_days_since_calibration"]]
print(filtered_df)

#Eksporterer filtrerte data til en ny JSON array 
with open(yml_data["output_file"], "w") as f:   
    filtered_df.to_json(f, orient="records", lines=True)  
```
